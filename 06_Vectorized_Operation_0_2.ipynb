```python
# A warehouse has 4 storage sections and 5 products.
# Initially, every product has a stock level of 100 units.
# Tasks

# Create a 4 × 5 array where every value is 100.

# Increase the stock of the first two sections by 20 units.

# Reduce stock of the last two products by 10%.

# Display only the middle two sections.

# Display only the first three products.

# Display the bottom-right 2 × 2 portion of the matrix.
```

```python
# Create a 4 × 5 array where every value is 100.
import numpy as np
wherehouse = np.full((4,5),100)
print(wherehouse)
```

```python
# Increase the stock of the first two sections by 20 units.
wherehouse[0:2]+=20
print(wherehouse)
```

```python
# Reduce stock of the last two products by 10%.
wherehouse[:,-2:]=wherehouse[:,-2:]*0.90
print(wherehouse)
```

```python
# Display only the middle two sections.
print(wherehouse[1:3,:])
```

```python
# Display only the first three products.
print(wherehouse[:,:3])
```

```python
wherehouse
```

```python
# Display the bottom-right 2 × 2 portion of the matrix.
print(wherehouse[-2:,-2:])
```

```python

# A startup expects its monthly users to grow from 10,000 to 100,000 over 12 months.

# Tasks

# Generate monthly user targets using np.linspace().

# Create month numbers using np.arange().

# Display users expected in Month 1 and Month 12.

# Display users expected from Months 4–8.

# Increase all targets by another 15%.

# Display the updated targets for the last 4 months.
```

```python
# 1. Generate monthly user targets using np.linspace()
# From 10,000 to 100,000 over 12 months
users = np.linspace(10000, 100000, 12)
print("Monthly user targets:")
print(users)
print()
```

```python
# 2. Create month numbers using np.arange()
months = np.arange(1, 13)
print("Month numbers:")
print(months)
print()
```

```python
# 3. Display users expected in Month 1 and Month 12
print("Users in Month 1 and Month 12:")
print(users[0], users[11])          # or users[[0, 11]]
print()
```

```python
# 4. Display users expected from Months 4–8
print("Users from Months 4 to 8:")
print(users[3:8])                   # indices 3 to 7 (Months 4 to 8)
print()
```

```python
# 5. Increase all targets by another 15%
users = users * 1.15
print("Updated targets after +15%:")
print(users)
print()
```

```python
# 6. Display the updated targets for the last 4 months
print("Updated targets for the last 4 months:")
print(users[-4:])
```

```python
# An e-commerce company stores order information.
# orders = np.array([
#     [1000, 2, 500],
#     [1001, 3, 750],
#     [1002, 1, 1200],
#     [1003, 4, 300],
#     [1004, 2, 900]
# ])

# Columns:

# Column 1 → Order ID
# Column 2 → Quantity
# Column 3 → Price per item

# Tasks

# Display the complete information of Order 1002.
# Display all quantities.
# Display all prices.
# Display the first 3 orders.
# Display the last 2 orders.
# Increase the price of the last 2 products by 10%.
# Increase the quantity of the first order by 1.
# Display the final dataset.
# Create a separate array containing only the quantity and price columns.
# Display the shape and size of the dataset
```

```python
orders = np.array([
    [1000, 2, 500],
    [1001, 3, 750],
    [1002, 1, 1200],
    [1003, 4, 300],
    [1004, 2, 900]
])
```

```python
# Display the complete information of Order 1002.
print(orders[2])
```

```python
# Display all quantities.
print(orders[:,1])
```

```python
# Display all prices.
print(orders[:,2])
```

```python
# Display the first 3 orders.
print(orders[:3])
```

```python
# Display the last 2 orders.
print(orders[-2:])
```

```python
# Increase the price of the last 2 products by 10%.
orders[-2:,2]=orders[-2:,2]*1.10
print(orders)
```

```python
# Increase the quantity of the first order by 1.
orders[0:1,1]+=1
print(orders)
```

```python
# Display the final dataset.
print(orders)
```

```python
# Create a separate array containing only the quantity and price columns.
print(orders[:,1:3])
```

```python
# Display the shape and size of the dataset.
print(orders.shape)
print(orders.size)
```
