```python
import numpy as np
# A bank account has the following transactions:
transactions = np.array([
5000, -2000, 15000, -4500, 7000,
-12000, 25000, -3000, 8000, -18000,
4500, 22000, -6000, 9000, -2500
])
# Positive values represent deposits and negative values represent withdrawals
```

```python
# Extract deposits.
# Extract withdrawals.
# Calculate total deposits and withdrawals.
# Calculate net transaction amount.
# Calculate absolute transaction values.
```

```python
# Extract deposits.
deposites = transactions[transactions > 0]
print(deposites)
```

```python
# Extract withdrawals.
withdrawals = transactions[transactions < 0]
print(withdrawals)
```

```python
# Calculate total deposits and withdrawals.
print(np.sum(deposites))
print(np.sum(withdrawals))
```

```python
# Calculate net transaction amount.
total_transactions = np.sum(transactions)
print(total_transactions)
```

```python
# Calculate absolute transaction values.
absolute = np.abs(transactions)
print(absolute)
```

```python
# Find transactions whose absolute value is above the average absolute transaction.
# Find the largest deposit.
# Find the largest withdrawal
# Create labels Credit and Debit
# Find percentage of transactions whose absolute value is greater than ₹10,000.
# Reshape the data into 3 × 5.
# Calculate total transaction value for each row
```

```python
# Find transactions whose absolute value is above the average absolute transaction.
avg_absolute = np.average(absolute)
absolute[absolute > avg_absolute]
```

```python
# Find the largest deposit.
lar_deposite = np.max(deposites)
print(lar_deposite)
```

```python
# Find the largest withdrawal
lar_withdrawl = np.max(withdrawals)
print(lar_withdrawl)
```

```python
# Create labels "Credit" and "Debit"
print(np.where(transactions > 0,"Credit","Debit" ))
```

```python
# Find percentage of transactions whose absolute value is greater than ₹10,000.
abs_t = np.abs(transactions)
abs10k = abs_t[abs_t > 10000]
print(abs10k)
```

```python
# A factory has four machines operating for six days:
# production = np.array([
# [120, 135, 128, 145, 150, 138],
# [110, 125, 118, 130, 142, 135],
# [160, 155, 170, 165, 180, 175],
# [95, 105, 100, 115, 108, 120]
#                         ])

# Calculate total production by machine.
# Calculate total production by day.
# Calculate average production of each machine.
# Find the machine with the highest average production.
# Find the day with the lowest production.
```

```python
production = np.array([
[120, 135, 128, 145, 150, 138],
[110, 125, 118, 130, 142, 135],
[160, 155, 170, 165, 180, 175],
[95, 105, 100, 115, 108, 120]
                        ])
```

```python
# Calculate total production by machine.
total_production = np.sum(production,axis=1)
print(total_production)
```

```python
# Calculate total production by day.
oneday_total_prd = np.sum(production,axis=0)
print(oneday_total_prd)
```

```python
# Calculate average production of each machine.
avg_prd = np.average(production,axis=1)
print(avg_prd)
```

```python
# Find the machine with the highest average production.
ind = np.argmax(avg_prd)
print(ind)
```

```python
# Find the day with the lowest production.
np.argmin(oneday_total_prd)
```

```python
# Find values above the overall average.
# Increase production values below 120 by 5%.
# Extract alternate days.
# Extract the last 3 days for machines 1 and 3.
# Create "High" and "Normal" labels based on the overall average
```

```python
# Find values above the overall average.
avg_production = np.average(production)
print(avg_production)
```

```python
# Increase production values below 120 by 5%.
production = np.where(production < 120 ,production*1.05,production)
print(production)
```

```python
# Extract alternate days.
print(production[:,::2])
```

```python
# Extract the last 3 days for machines 1 and 3.
print(production[[0,2],-3:])
```

```python
# Create "High" and "Normal" labels based on the overall average
np.where(production > avg_production,"High","Normal")
```

```python
# Consider the following loan portfolio:
# loan_amount = np.array([
# 100000, 250000, 500000, 750000,
# 120000, 350000, 900000, 450000
# ])
# interest_rate = np.array([
# 8.5, 9.2, 8.8, 10.1,
# 7.9, 9.5, 10.5, 8.7
# ])
# Calculate annual interest for every loan.
# Calculate total interest.
# Calculate average interest.
# Find the loan generating the highest interest.
# Find loans above ₹400,000 with interest rate above 9%
```

```python
loan_amount = np.array([
100000, 250000, 500000, 750000,
120000, 350000, 900000, 450000
])
interest_rate = np.array([
8.5, 9.2, 8.8, 10.1,
7.9, 9.5, 10.5, 8.7
])
```

```python
# Calculate annual interest for every loan.
annual_intrest = loan_amount*interest_rate/100
print(annual_intrest)
```

```python
# Calculate total interest.
total_intrest = np.sum(annual_intrest)
print(total_intrest)
```

```python
# Calculate average interest.
avg_intrest = np.average(annual_intrest)
print(avg_intrest)
```

```python
# Find the loan generating the highest interest.
high_ind = np.argmax(annual_intrest)
print(high_ind)
```

```python
# Find loans above ₹400,000 with interest rate above 9%
critaria = (loan_amount > 400000) & (interest_rate > 9)
print(critaria)
loan_amount[critaria]
```
