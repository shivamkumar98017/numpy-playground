```python
campaigns = np.array([
    [101, 10000, 800, 80, 5000, 7500],
    [102, 15000, 1200, 150, 7000, 12000],
    [103, 8000, 400, 30, 3000, 4200],
    [104, 20000, 1800, 250, 9000, 18000],
    [105, 12000, 600, 90, 6000, 8000],
    [106, 18000, 1500, 220, 8000, 16000],
    [107, 9000, 300, 20, 2500, 3000],
    [108, 25000, 2200, 300, 11000, 24000]
])
```

```python
# Campaign_ID
# Impressions
# Clicks
# Conversions
# Cost
# Revenue
```

```python
# CTR = Clicks / Impressions * 100

# Conversion Rate = Conversions / Clicks * 100

# ROI = (Revenue - Cost) / Cost * 100
```

```python
# Find highest-revenue campaign.
# Find highest-ROI campaign.
# Calculate average CTR.
# Calculate average ROI.
# Find campaigns with above-average CTR and ROI.
```

```python
# Find highest-revenue campaign.
Revenue = campaigns[:,-1]
Revenue
np.argmax(Revenue)
campaigns[7]
```

```python
Cost = campaigns[:,-2]
Cost
```

```python
# Find highest-ROI campaign.
ROI = (Revenue - Cost) / Cost * 100
ROI
```

```python
Clicks = campaigns[:,2]
Impressions = campaigns[:,1]
```

```python
# Calculate average CTR.
CTR = Clicks / Impressions * 100
CTR
```

```python
avg_CTR = np.mean(CTR)
avg_CTR
```

```python
# Calculate average ROI.
avg_ROI = np.mean(ROI)
avg_ROI
```

```python
# Find campaigns with above-average CTR and ROI.
Criteria = (CTR > avg_CTR) & (ROI > avg_ROI)
campaigns[Criteria]
```

```python
# # Categorize campaigns:
# ROI >= 80 --> Excellent
# ROI >= 40 --> Good
# otherwise --> poor

categorize = np.where(ROI >= 80,"Excellent",np.where(ROI >= 40,"Good","Poor"))
print(categorize)
```

```python
# Calculate percentage of excellent campaigns.
m = len(categorize[categorize == "Excellent"])
n = len(categorize)
perc = m*100/n
print(perc)
```
