```python
import numpy as np
employees = np.array([
[101, 45000, 120000],
[102, 60000, 180000],
[103, 35000, 90000],
[104, 75000, 250000],
[105, 50000, 160000]
])
sales = employees[:,2]
# 2. Sales > 200000, excellent, 150000 --> good, else improvement.
category = np.where(sales > 200000, "Excellent", np.where(sales>150000,"Good","Improvement"))
print(category)
```

```python
#  An e-commerce company records the quantity sold and selling price of 12 products.
quantity = np.array([2, 5, 3, 8, 4, 6, 10, 1, 7, 5, 9, 3])
price = np.array([
1200, 800, 1500, 450, 2200, 900,
700, 1800, 650, 1300, 500, 2500
])
```

```python
# Calculate revenue for every product.
# Calculate the total revenue.
# Find the product index generating the highest revenue.
# Extract the revenue of the last five products.
```

```python
# Calculate revenue for every product.
revenue = quantity*price
print(revenue)
```

```python
# Calculate the total revenue.
print(np.sum(revenue))
```

```python
# Find the product index generating the highest revenue.
np.argmax(revenue)
```

```python
# Extract the revenue of the last five products.
print(revenue[-5:])
```

```python
# Find products whose revenue is above the average revenue.
# Apply a 10% discount only to products whose price is greater than ₹1,000.
# Round the discounted prices to two decimal places.
# Calculate the total quantity sold.
# Find the median product price.
# Identify products whose price is above the median and whose quantity is at least 5.
```

```python
# Find products whose revenue is above the average revenue.
revenue[revenue > np.mean(revenue)]
```

```python
np.where(revenue > np.mean(revenue),"Yes","No")
```

```python
# Apply a 10% discount only to products whose price is greater than ₹1,000.
discount_price = np.where(price>1000,price*0.9,price)
print(discount_price)
```

```python
# Round the discounted prices to two decimal places.
discount_price = np.round(discount_price,2)
print(discount_price)
```

```python
# Calculate the total quantity sold.
np.sum(quantity)
```

```python
# Find the median product price.
median = np.median(price)
print(median)
```

```python
# Identify products whose price is above the median and whose quantity is at least 5.
prds =(price > median) & (quantity >= 5)
price[prds]
```

```python
criteria = (price > median) & (quantity >= 5)
price[criteria]
```

```python
# A company's monthly sales for one year are:
sales = np.array([
120000, 135000, 128000, 150000,
160000, 172000, 165000, 180000,
195000, 210000, 205000, 225000
])
```

```python
# Calculate the average monthly sales.
# Find the median sales.
# Find the month with the highest sales.
# Calculate the standard deviation
```

```python
# Calculate the average monthly sales.
np.average(sales)
```

```python
# Find the median sales.
np.median(sales)
```

```python
# Find the month with the highest sales.
np.argmax(sales)
```

```python
# Calculate the standard deviation
np.std(sales)
```

```python
# Create a forecast for the next 6 months using np.linspace() between:
# the minimum observed sales
# and 20% above the maximum observed sales.
# Round the forecast values to the nearest rupee.
# Combine the original sales and forecast into one 18-element array.
# Reshape the combined array into 3 × 6.
# Calculate the average sales for each row.
# Identify values that are greater than the overall median.
```

```python
mn_s = np.min(sales)
mx_s = np.max(sales)
mx_s20 = mx_s*1.20
```

```python
forecast = np.linspace(mn_s,mx_s20,6)
forecast
```

```python
# Round the forecast values to the nearest rupee.
r_s = np.round(forecast,2)
```

```python
r_s
```

```python
# Combine the original sales and forecast into one 18-element array.
n_s = np.append(sales,r_s)
```

```python
n_s
```

```python
# Reshape the combined array into 3 × 6.
ans = n_s.reshape(3,6)
ans
```

```python
# Calculate the average sales for each row.
np.mean(ans,axis=1)
```

```python
# Identify values that are greater than the overall median.
np.where(n_s > np.median(n_s),"Yes","No")
```
