```python
# An electricity board records the monthly electricity consumption of
#  a household for one year:

# consumption = np.array([
#     180, 220, 195, 250,
#     280, 310, 295, 270,
#     240, 210, 190, 230
# ])

# Perform the following:

# 1. Find the total annual consumption.
# 2. Calculate average monthly consumption.
# 3. Find the highest and lowest consumption.
# 4. Find the month index with the highest consumption.
# 5. Extract the first 6 months using slicing.
# 6. Extract alternate months.
# 7. Find all months where consumption was greater than the annual average.
# 8. Calculate the percentage increase required if next year's consumption is
#  expected to be 8% higher.
```

```python
consumption = np.array([
     180, 220, 195, 250,
     280, 310, 295, 270,
     240, 210, 190, 230
 ])
```

```python
# 1. Find the total annual consumption.
print(np.sum(consumption))
```

```python
# 2. Calculate average monthly consumption.
print(np.average(consumption))
```

```python
# 3. Find the highest and lowest consumption.
print(np.max(consumption))
print(np.min(consumption))
```

```python
# 4. Find the month index with the highest consumption.
print(np.argmax(consumption))
```

```python
# 5. Extract the first 6 months using slicing.
print(consumption[0:6])
```

```python
# 6. Extract alternate months.
print(consumption[0:13:2])
```

```python
# 7. Find all months where consumption was greater than the annual average.
consumption[consumption > np.average(consumption)]
```

```python
# 8. Calculate the percentage increase required if next year's consumption is expected to be 8% higher.
consumption = consumption*1.08
print(consumption)
```

```python
# The following array contains marks of 5 students in 4 subjects:

# marks = np.array([
#     [72, 85, 90, 78],
#     [65, 70, 75, 80],
#     [88, 92, 95, 90],
#     [55, 60, 58, 65],
#     [78, 82, 85, 88]
# ])

# Assume each row represents a student and each column represents a subject.

# Find:

# 1. Shape and size of the array.
# 2. Marks obtained by the third student.
# 3. Marks obtained in the second subject by all students.
# 4. Average marks of every student.
# 5. Average marks of every subject.
# 6. Students whose average is above 75.
# 7. Highest mark obtained in each subject.
# 8. The complete row of the student with the highest average.
```

```python
marks = np.array([
    [72, 85, 90, 78],
    [65, 70, 75, 80],
    [88, 92, 95, 90],
    [55, 60, 58, 65],
    [78, 82, 85, 88]
])
```

```python
# 1. Shape and size of the array.
print(np.shape(marks))
print(np.size(marks))
```

```python
# 2. Marks obtained by the third student.
print(marks[2])
```

```python
# 3. Marks obtained in the second subject by all students.
print(marks[:,1])
```

```python
# 4. Average marks of every student.
print(np.average(marks, axis=1))
```

```python
# 5. Average marks of every subject.
print(np.average(marks, axis=0))
```

```python
# 6. Students whose average is above 75.
marks[75<np.average(marks, axis=1)]
```

```python
# 7. Highest mark obtained in each subject.
print(np.max(marks, axis=0))
```

```python
# 8. The complete row of the student with the highest average.
print(np.argmax(np.average(marks, axis= 1)))
```

```python
# An online store has the following product prices:

# prices = np.array([
#     499.50, 799.90, 1299.75,
#     2499.50, 3999.90, 599.25,
#     899.80, 1599.40
# ])

# Perform:

# 1. Round prices to the nearest whole number.
# 2. Find products costing more than ₹1,000.
# 3. Calculate a 15% discount only on products costing more than ₹1,000.
# 4. Calculate the average discounted price of those products.
# 5. Find the most expensive original product.
# 6. Extract the last 4 products.
# 7. Calculate the square root of the final prices and round them to 2 decimal places.
```

```python
prices = np.array([
    499.50, 799.90, 1299.75,
    2499.50, 3999.90, 599.25,
    899.80, 1599.40
])
```

```python
# 1. Round prices to the nearest whole number.
np.round(prices)
```

```python
# 2. Find products costing more than ₹1,000.
prices[prices>1000]
```

```python
# 3. Calculate a 15% discount only on products costing more than ₹1,000.
prices = prices[prices>1000]*0.85
print(prices)
```

```python
# 4. Calculate the average discounted price of those products.
print(np.average(prices))
```

```python
# 5. Find the most expensive original product.
print(np.max(prices))
```

```python
np.argmax(prices)
```

```python
# 6. Extract the last 4 products.
print(prices[-4:])
```

```python
# 7. Calculate the square root of the final prices and round them to 2 decimal places.
sqrt_prices = np.sqrt(prices)
final = np.round(sqrt_prices,2)
print(final)
```
