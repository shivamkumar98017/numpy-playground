```python
# # A factory collects temperatures from 20 sensor readings:
# temperature = np.array([
# 28.5, 31.2, 29.8, 35.6, 38.1,
# 40.2, 27.9, 30.5, 42.3, 39.8,
# 36.4, 33.2, 29.1, 41.7, 44.2,
# 31.8, 34.5, 37.6, 26.9, 43.1
# ])
# Calculate mean, median, minimum and maximum.
# Calculate variance and standard deviation.
# Find all temperatures more than one standard deviation above the mean.
# Calculate the absolute difference between every temperature and the mean.
# Find the five readings having the largest deviation from the mean.
# Round all temperatures to the nearest whole number.
```

```python
import numpy as np
temperature = np.array([
28.5, 31.2, 29.8, 35.6, 38.1,
40.2, 27.9, 30.5, 42.3, 39.8,
36.4, 33.2, 29.1, 41.7, 44.2,
31.8, 34.5, 37.6, 26.9, 43.1
])
```

```python
temperature.ndim
```

```python
temperature.shape
```

```python
temperature.size
```

```python
# Calculate mean, median, minimum and maximum.
temp_mean = np.mean(temperature)
temp_mean
```

```python
mediant = np.median(temperature)
mediant
```

```python
temp_min = np.min(temperature)
temp_mean
```

```python
temp_max = np.max(temperature)
temp_max
```

```python
# Calculate variance and standard deviation.
std_temp = np.std(temperature)
std_temp
```

```python
var = np.square(std_temp)
var
```

```python
# Find all temperatures more than one standard deviation above the mean.
new_t = temperature[temperature >(temp_mean + std_temp) ]
new_t
```

```python
np.where(temperature > (temp_mean + std_temp),"Yes","No")
```

```python
# Calculate the absolute difference between every temperature and the mean.
deviation = np.abs(temperature-temp_mean)
deviation
```

```python
# Find the five readings having the largest deviation from the mean.
largest_deviation  = np.sort(deviation)
largest_deviation
```

```python
largest_deviation[-5:]
```

```python
# Round all temperatures to the nearest whole number.
np.round(temperature,2)
```
