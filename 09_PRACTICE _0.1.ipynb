```python
marks = np.array([45,72,88,31,65,90])
# marks > 70--> "Good", else "Average"
result = np.where(marks >70, "Good", "Average")
print(result)
```

```python
marks = np.array([45,72,88,31,65,90])
# Suppose students scoring below 50 get 5 grace marks.
new_marks = np.where(marks < 50, marks+5, marks )
print(new_marks)
```

```python
# Suppose products costing more than ₹1,000 receive a 10% discount.
prices = np.array([500, 1200, 800, 2500, 1500])
new_prices = np.where(prices>1000, prices*0.90, prices )
print(new_prices)
```

```python
salary = np.array([25000, 45000, 75000, 32000, 90000])
# Salary >= 50000 → High
# Salary < 50000 → Standard
n_salary = np.where(salary >= 50000, "High","Standard" )
print(n_salary)
```

```python
marks = np.array([45,72,88,31,65,90])
# 90 or above → Excellent
# 50–89 → Pass
# Below 50 → Fail
result_marks = np.where(marks>=90,"Excellent",np.where(marks>50, "Pass", "Fail"))
print(result_marks)
```

```python
# Suppose a customer receives free shipping if:
# order value >= ₹1,000
# OR
# customer is a premium customer.
# order_value = np.array([500, 1200, 700, 800, 1500])
# premium = np.array([False, False, True, False, True])
```

```python
# Consider:
# employees = np.array([
# [101, 35000, 2],
# [102, 48000, 4],
# [103, 62000, 6],
# [104, 40000, 3],
# [105, 75000, 8],
# [106, 55000, 5]
# ])
# Columns represent:
# Employee ID | Salary | Experience
# Find:
# Employees earning more than ₹50,000.
# Employees with more than 4 years of experience.
# Employees earning more than ₹50,000 AND having more than 4 years of experience.
# Average salary of employees satisfying condition 3.
# Employee with the highest salary.
# Employees earning below the median salary.
```

```python
employees = np.array([
[101, 35000, 2],
[102, 48000, 4],
[103, 62000, 6],
[104, 40000, 3],
[105, 75000, 8],
[106, 55000, 5]
])
```

```python
# Employees earning more than ₹50,000.
print(employees[employees[:,1]>50000])
```

```python
# Employees with more than 4 years of experience.
print(employees[employees[:,2]>4])
```

```python
# Employees earning more than ₹50,000 AND having more than 4 years of experience.
print(employees[employees[:,1]>50000]& employees[employees[:,2]>4])
```

```python
# Average salary of employees satisfying condition 3.
cond3 = employees[employees[:,1]>50000]& employees[employees[:,2]>4]
print(np.average(cond3))
```

```python
# Employee with the highest salary.
highest = employees[employees[:,1].argmax()]
print(highest)
```

```python
# Employees earning below the median salary.
median_salary = np.median(employees[:,1])
below_median = employees[employees[:,1] < median_salary]
print("Median Salary:", median_salary)
print(below_median)
```
