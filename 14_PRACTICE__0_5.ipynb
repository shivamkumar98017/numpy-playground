```python
# # Inventory across four stores and five products:

# inventory = np.array([
#     [12, 45, 8, 32, 20],
#     [5, 60, 15, 25, 10],
#     [30, 18, 6, 40, 22],
#     [9, 55, 12, 28, 7]
# ])

# # Tasks:
# # Calculate total inventory per store.
# # Calculate total inventory per product.
# # Find the store with highest inventory.
# # Find the product with lowest total inventory.
# # Extract inventory values below 10.
```

```python
import numpy as np
inventory = np.array([
    [12, 45, 8, 32, 20],
    [5, 60, 15, 25, 10],
    [30, 18, 6, 40, 22],
    [9, 55, 12, 28, 7]
])
```

```python
# Calculate total inventory per store.
total_inventory_store = np.sum(inventory,axis=1)
print(total_inventory_store)
```

```python
# Calculate total inventory per product.
total_inventory_product = np.sum(inventory,axis=0)
print(total_inventory_product)
```

```python
# Find the store with highest inventory.
highest_inventory = np.argmax(total_inventory_store)
print(highest_inventory+1)
```

```python
# Find the product with lowest total inventory.
lowest_inventory = np.argmin(total_inventory_product)
print(lowest_inventory+1)
```

```python
# Extract inventory values below 10.
low_inventory = inventory[inventory < 10]
print(low_inventory)
```

```python
# Count items that require replenishment.
# Create "Reorder" / "OK" labels.
# Increase inventory below 10 by 20%.
# Round the resulting inventory.
# Find median inventory.
# Extract the last two products.
# Flatten the inventory and calculate standard deviation.
```

```python
# Count items that require replenishment.(<10)
n = len(low_inventory)
print(n)
```

```python
m = np.sum(inventory < 10)
print(m)
```

```python
# Create "Reorder" / "OK" labels.
label = np.where(inventory < 10, "Reorder","Ok")
print(label)
```

```python
# Increase inventory below 10 by 20%.
new_inv = np.where(inventory < 10, inventory*1.2, inventory )
print(new_inv)
```

```python
# Round the resulting inventory.
round = np.round(new_inv)
print(round)
```

```python
# Find median inventory.
np.median(new_inv)
```

```python
# Consider the following daily prices:

prices = np.array([
    820, 835, 810, 850, 875,
    860, 890, 910, 895, 930,
    945, 920, 960, 975, 990
])

# Calculate the change between consecutive days.
# Calculate absolute price movement.
# Find the largest movement.
# Find average movement.
# Find movements above average.
# Calculate percentage change between consecutive days.
# Round percentage changes to 2 decimals.
# Find days where price increased by more than 2%.
# Find maximum, minimum and range.
# Reshape prices into 3 x 5.
# Find average price of each row.
```

```python
prices = np.array([
    820, 835, 810, 850, 875,
    860, 890, 910, 895, 930,
    945, 920, 960, 975, 990
])
```

```python
# Calculate the change between consecutive days.
a = prices[1:]
print(a)
```

```python
b = prices[:-1]
print(b)
```

```python
ans = a-b
print(ans)
```

```python
# Calculate absolute price movement.
abs_price = np.abs(ans)
print(abs_price)
```

```python
# Find the largest movement.
largest_movement = np.max(ans)
print(largest_movement)
```

```python
# Find average movement.
avg_mov = np.average(ans)
print(avg_mov)
```

```python
# Find movements above average.
ans[ans > avg_mov]
```

```python
# Calculate percentage change between consecutive days.
percentage = (ans / b) * 100
print("Percentage change:", percentage)
```

```python
# Round percentage changes to 2 decimals.
percentage = np.round(percentage, 2)
print("Rounded percentage:", percentage)
```

```python
# Find days where price increased by more than 2%.
above_2 = percentage > 2
print("Increase above 2%:", above_2)
```

```python
# Find maximum, minimum and range.
maximum = np.max(prices)
minimum = np.min(prices)
price_range = maximum - minimum
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Range:", price_range)
```

```python
# Reshape prices into 3 x 5.
prices_3x5 = prices.reshape(3, 5)
print("3 x 5:")
print(prices_3x5)
```

```python
# Find average price of each row.
row_average = np.mean(prices_3x5, axis=1)
print("Row averages:", row_average)
```
