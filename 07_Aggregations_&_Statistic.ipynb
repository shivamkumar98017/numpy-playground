```python
# A company stores quarterly revenue and expenses:

# financials = np.array([
#     [120, 80],
#     [150, 90],
#     [180, 110],
#     [200, 130]
# ])

# Rows = Q1, Q2, Q3, Q4
# Columns = Revenue, Expense

# Tasks
# Display Q2's complete financial data.
# Display all quarterly revenues.
# Display all quarterly expenses.
# Extract Q2 and Q3 data.
# Increase Q4 revenue by 15%.
# Reduce Q1 expense by 10%.
# Display the final financial matrix.
```

```python
financials = np.array([
    [120, 80],
    [150, 90],
    [180, 110],
    [200, 130]
])
```

```python
# Display Q2's complete financial data.
print(financials[1])
```

```python
# Display all quarterly revenues.
print(financials[:,0])
```

```python
# Display all quarterly expenses.
print(financials[:,1])
```

```python
# Extract Q2 and Q3 data.
print(financials[1:3])
```

```python
# Increase Q4 revenue by 15%.
financials[3:,0]=financials[3:,0]*1.15
print(financials)
```

```python
# Reduce Q1 expense by 10%.
financials[0:,1]=financials[0:,1]*0.9
print(financials)
```

```python
# Display the final financial matrix.
print(financials)
```

```python
# A classroom has 6 rows and 5 seats per row.

# Initially, all seats are empty.

# 0 = Empty
# 1 = Occupied

# Tasks
# Create a 6 x 5 array filled with zeros.
# Mark the first seat of the first row as occupied.
# Mark all seats in the 3rd row as occupied.
# Mark the last two seats of the last row as occupied.
# Display the first 3 rows.
# Display the last 2 columns.
# Display the complete seating matrix.
```

```python
# Create a 6 x 5 array filled with zeros.
seats =np.zeros((6,5),dtype=int)
print(seats)
```

```python
# Mark the first seat of the first row as occupied.
seats[0,0]=1
print(seats)
```

```python
# Mark all seats in the 3rd row as occupied.
seats[2]= 1
print(seats)
```

```python
# Mark the last two seats of the last row as occupied.
seats[-1,-2:]=1
print(seats)
```

```python
# Display the first 3 rows.
print(seats[:3])
```

```python
# Display the last 2 columns.
print(seats[:,-2:])
```

```python
# Display the complete seating matrix.
print(seats)
```

```python
# A teacher has recorded marks of 10 students:

# marks = np.array([72, 85, 91, 64, 78, 88, 95, 70, 82, 76])

# Find:

# Total marks
# Average marks
# Highest marks
# Lowest marks
# Difference between highest and lowest marks
# Index of the student with the highest marks
```

```python
marks = np.array([72, 85, 91, 64, 78, 88, 95, 70, 82, 76])
```

```python
# Total marks
print(np.sum(marks))
```

```python
# Average marks
print(np.average(marks))
```

```python
# Highest marks
print(np.max(marks))
```

```python
# Lowest marks
print(np.min(marks))
```

```python
# Difference between highest and lowest marks.
print(np.max(marks)-np.min(marks))
```

```python
# Index of the student with the highest marks.
print(np.argmax(marks))
```

```python
# A company's daily sales for 7 days are:

# sales = np.array([45000, 47000, 49000, 52000, 50000, 48000, 250000])

# Calculate the mean and median.

# Then determine which measure better represents a typical sales day when considering the unusually large sale.

# Also calculate the standard deviation.
```

```python
sales = np.array([45000, 47000, 49000, 52000, 50000, 48000, 250000])
```

```python
# Calculate the mean and median.
print(np.mean(sales))
print(np.median(sales))
```

```python
# Also calculate the standard deviation.
print("Standard deviation:", np.std(sales))
```
