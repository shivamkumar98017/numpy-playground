```python
import numpy as np

loan_amount = np.array([
    100000, 250000, 500000, 750000,
    120000, 350000, 900000, 450000
])

interest_rate = np.array([
    8.5, 9.2, 8.8, 10.1,
    7.9, 9.5, 10.5, 8.7
])
```

```python
# Categorize loans:
# Rate >= 10 → High
# Rate >= 9 → Medium
# Otherwise → Low
categories = np.where(interest_rate >= 10,"High",np.where(interest_rate>=9,"Medium","Low"))
print(categories)
```

```python
# Round interest to 2 decimal places.
intrest = loan_amount*interest_rate/100
round = np.round(intrest,2)
print(round)
```

```python
# Find median loan amount.
median = np.median(loan_amount)
print(median)
```

```python
# Find loans above median amount AND interest above average interest.
avg_intrest = np.average(interest_rate)
criteria = (loan_amount > median) & (interest_rate >avg_intrest)
loan_amount[criteria]
```

```python
# Calculate standard deviation of intrest rates.
std_dev = np.std(interest_rate)
print(std_dev)
```

```python
# The coordinates of eight delivery locations are:
# points = np.array([
# [3, 4],
# [5, 12],
# [8, 15],
# [7, 24],
# [9, 40],
# [12, 35],
# [15, 20],
# [6, 8]
# ])

# Each row contains:

# [x, y]

# Calculate distance from the origin using:

# distance = √(x² + y²)

# Extract x and y coordinates.

# Calculate distance for every location.

# Round distances to 2 decimals.

# Find nearest and farthest location.

# Find average distance.

# Find locations above average distance.

# Categorize:
# if >= 30 → Long
# elif >= 15 → Medium
# otherwise --> short

# Find percentage of Long-Distance delivery.

# Flatten the coordinate array.

# Find the size of flatten array.
```

```python
points = np.array([
[3, 4],
[5, 12],
[8, 15],
[7, 24],
[9, 40],
[12, 35],
[15, 20],
[6, 8]
])
```

```python
# Calculate distance from the origin using:
# distance = √(x² + y²)
```

```python
# Extract x and y coordinates.
x = points[:,0]
y = points[:,1]
print("x = ",x)
print("y = ",y)
```

```python
# Calculate distance for every location.
distances = np.sqrt(np.square(x)+np.square(y))
distances
```

```python
# Round distances to 2 decimals.
round_dist = np.round(distances)
print(round_dist)
```

```python
# Find nearest and farthest location.
nearest_location = np.min(distances)
print(nearest_location)
```

```python
# Find average distance.
avg_distance = np.average(distances)
print(avg_distance)
```

```python
# Find locations above average distance.
points[distances > avg_distance]
```

```python
# Categorize:
# if >= 30 → Long
# elif >= 15 → Medium
# otherwise --> short
categorize = np.where(distances >= 30, "Long",np.where(distances >= 15,"Medium","short"))
print(categorize)
```

```python
# Find percentage of Long-Distance delivery.
long_dist = categorize[categorize == "Long"]
n = len(distances)
m = len(long_dist)
perc = (m*100)/n
print(perc)
```

```python
# Flatten the coordinate array.
flatten_array = points.flatten()
print(flatten_array)
```

```python
# Find the size of flatten array.
n = np.size(flatten_array)
print(" Size of Flatten array = ",n)
```

```python
patients = np.array([
    [101, 120, 85, 98],
    [102, 135, 92, 96],
    [103, 110, 78, 99],
    [104, 150, 88, 94],
    [105, 125, 95, 97],
    [106, 160, 82, 93],
    [107, 115, 90, 98],
    [108, 145, 87, 95],
    [109, 130, 91, 96],
    [110, 155, 80, 92]
])
```

```python
# Calculate average heart rate.
avg_heart_rate = np.average(patients[:,1])
print(avg_heart_rate)
```

```python
# Calculate median oxygen level.
median_oxy_lvl = np.median(patients[:,2])
print(median_oxy_lvl)
```

```python
# Find patients with heart rate above average AND oxygen below 90.
criterias = (patients[:,1]>avg_heart_rate) & (patients[:,2] < 90)
print(criterias)
patients[criterias]
```

```python
# Categorize recovery:
# >= 97 → Excellent
# >= 95 → Stable
# Otherwise → Monitor

# Count each category.

# Find patient with highest heart rate.

# Calculate standard deviation of oxygen levels.

# Find oxygen levels more than 1 standard deviation below the mean.

# Extract the last 3 patients.

# Flatten the measurement data and calculate its mean.
```

```python
# Count each category.
count_excellent = len(Categorize_recovery[Categorize_recovery == "Excellent"])
count_suitable = len(Categorize_recovery[Categorize_recovery == "Suitable"])
count_monitor = len(Categorize_recovery[Categorize_recovery == "Monitor"])
print(count_excellent)
print(count_suitable)
print(count_monitor)
```
