```python
import numpy as np

prices = np.array([
    820, 835, 810, 850, 875,
    860, 890, 910, 895, 930,
    945, 920, 960, 975, 990
])

# Calculate percentage change between consecutive days.
# Round percentage changes to 2 decimals.
# Find days where price increased by more than 2%.
# Find maximum, minimum and range.
# Reshape prices into 3 x 5.
# Find average price of each row.
```

```python
# Calculate percentage change between consecutive days.
diff = prices[1:]-prices[:-1]
print(diff)
```

```python
perc_change = (diff*100)/prices[:1]
print(perc_change)
```

```python
# Round percentage changes to 2 decimals.
round_perc = np.round(perc_change,2)
print(round_perc)
```

```python
# Find days where price increased by more than 2%.
high_change_day = np.where(perc_change > 2,"Yes","No")
print(high_change_day)
```

```python
# Find maximum, minimum and range.
max_change = np.max(perc_change)
min_change = np.min(perc_change)
range_perc_change = max_change - min_change

print("max_change:", max_change)
print("min_change:", min_change)
print("range:", range_perc_change)
```

```python
# Reshape prices into 3 x 5.
new_prices = prices.reshape((3,5))
print(new_prices)
```

```python
# Find average price of each row.
avg_price = np.mean(new_prices,axis=1)
print(avg_price)
```

```python
# A company records leads generated by four campaigns across five channels:

leads = np.array([
    [12000, 15000, 11000, 18000, 21000],
    [14000, 13000, 17000, 19000, 22000],
    [10000, 16000, 15000, 20000, 23000],
    [18000, 17000, 19000, 21000, 25000]
])

# Find total leads generated by each campaign.
# Find total leads generated by each channel.
# Find the campaign with maximum leads.
# Find the channel with maximum leads.
```

```python
# Find total leads generated by each campaign.
total_each_compain = np.sum(leads,axis=1)
print(total_each_compain)
```

```python
# Find total leads generated by each channel.
total_each_channel = np.sum(leads,axis=0)
print(total_each_channel)
```

```python
# Find the campaign with maximum leads.
compain_id = np.argmax(total_each_compain)+1
print(compain_id)
```

```python
# Find the channel with maximum leads.
channel_id= np.argmax(total_each_channel)+1
print(channel_id)
```

```python
# Find average leads per campaign.
# Find values above overall average.
# Increase values below 15,000 by 10%.
# Round the resulting values.
# Categorize values above median as "High" and others as "Low".
# Extract every second channel.
```

```python
# Find average leads per campaign.
avg_compain = np.mean(total_each_compain)
print(avg_compain)
```

```python
overall_avg = np.mean(leads)
print(overall_avg)
```

```python
# Find values above overall average.

high_leads = leads[leads > overall_avg]
print(high_leads)
```

```python
# Increase values below 15,000 by 10%.
new_leads = np.where(leads< 15000,leads*1.10,leads)
print(new_leads)
```

```python
# Round the resulting values.
round_result = np.round(new_leads)
print(round_result)
```

```python
# Categorize values above median as "High" and others as "Low".
median_lead = np.median(leads)
caterize = np.where(leads > median_lead,"High","Low" )
print(caterize)
```

```python
# Extract every second channel.
channels = leads[:,1::2]
print(channels)
```

```python
# Consider the following e-commerce dataset:

orders = np.array([
    [101, 3, 1200, 4],
    [102, 5, 800, 3],
    [103, 2, 2500, 4],
    [104, 8, 450, 4],
    [105, 4, 1800, 3],
    [106, 6, 950, 4],
    [107, 3, 2200, 4],
    [108, 10, 600, 3],
    [109, 7, 1500, 4],
    [110, 2, 3000, 4],
    [111, 5, 750, 3],
    [112, 9, 1250, 4]
])

# Order_ID
# Quantity
# Price
# Rating
```

```python
# Inspect dimensions, shape and size.
# Extract the first 5 orders.
# Extract the last 3 orders.
# Extract every second order.
# Calculate revenue for every order.
```

```python
# Inspect dimensions, shape and size.
orders_shape = np.shape(orders)
print(orders_shape)
```

```python
orders_dimension = np.ndim(orders)
print(orders_dimension)
```

```python
orders_size = np.size(orders)
print(orders_size)
```

```python
# Extract the first 5 orders.
orders[:5]
```

```python
# Extract the last 3 orders.
orders[-3:]
```

```python
# Extract every second order.
orders[1::2]
```

```python
# Calculate revenue for every order.
quantity = orders[:,1]
price = orders[:,2]
revenue = quantity * price
print(revenue)
```

```python
# calculate total and average revenue.
# Find the order with highest revenue.
# Calculate median revenue.
```

```python
# calculate total and average revenue.
total_revenue = np.sum(revenue)
avg_revenue = np.mean(revenue)
print(total_revenue)
print(avg_revenue)
```

```python
# Find the order with highest revenue.
orders[np.argmax(revenue)]
```

```python
# Calculate median revenue.
median_revenue = np.median(revenue)
print(median_revenue)
```

```python
# Categorize orders:
# Revenue >= 15,000 → High
# Revenue >= 7,500 → Medium
# Otherwise → Low

# Find orders where quantity >= 5 AND rating >= 4.5.

# Find orders where price is above median OR rating is below 4.

# Give 10% discount to products whose price is above ₹1,000.
```

```python
# Categorize orders:
# Revenue >= 15,000 → High
# Revenue >= 7,500 → Medium
# Otherwise → Low

criteria = np.where(revenue>= 15000,"High",np.where(revenue >= 7500,"Medium","Low"))
print(criteria)
```

```python
# Find orders where quantity >= 5 AND rating >= 4.5.
rating = orders[:, -1]
high_order = (quantity >= 5) & (rating >= 4.5)
print(high_order)
orders[high_order]
```

```python
# Find orders where price is above median OR rating is below 4.
median_sales = np.median(sales)
new_orders = (price > median_sales) | (rating < 4)
print(new_orders)
orders[new_orders]
```

```python
# Give 10% discount to products whose price is above ₹1,000.
discount_10 = np.where(price > 10000, price*0.9, price )
print(discount_10)
```
